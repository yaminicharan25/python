```python
#list in python
#list is an ordered and changeable collection that can store multiple values
marks = [90, 80, 70, 60, 50]

print(marks) 
```

```python
#accessing elements in a list
marks = [90, 80, 70, 60, 50]

print(marks[0]) 
print(marks[1])
print(marks[2])
print(marks[3])
print(marks[4])
```

```python
#accessing elements in a list
marks = [29,64,36,991,44,79]

print(marks[0]) 
print(marks[1])
print(marks[2])
print(marks[3])
print(marks[4])
print(marks[5])
```

```python
#changing elements in a list
marks = [79,64,92,74,96,77,94]

marks[1] = 100
print(marks)

marks[3] = 88
print(marks)    

marks[5] = 99
print(marks)
```

```python
#index

numbers = [10, 20,30,40]

print(numbers.index(30))
```

```python
#sort

numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)
```

```python
# reverse
numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)

numbers.sort(reverse=True)

print(numbers)
```

```python
#copy method

a = [1, 2, 3]

b = a.copy()

print(b)
```
